```python
metadata_path_csv = "../dataset/polar/polar_annotation.csv"
df = pd.read_csv(metadata_path_csv)
```

```python
#show unique patient_id values count
print("Unique patient_id values:")
print(len(df['patient_id'].unique()))
print("\n")

#show unique polyp_id values count
print("Unique polyp_id values:")
print(len(df['polyp_id'].unique()))
print("\n")

#show unique img_file_prefix values
print("Unique image_file_prefix values:")
print(len(df['img_file_prefix'].unique()))
print("\n")

# create freq table for size and than pathology_diagnosis
print("Frequency table for size:")
print(df['size'].value_counts()/ df.shape[0] * 100 )
print("\n")

print("\nFrequency table for pathology_diagnosis:")
print(df['pathology_diagnosis'].value_counts()/ df.shape[0] * 100 )
print("\n")
```
