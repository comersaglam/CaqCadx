```python
import pandas as pd
from pathlib import Path
metadata_path_csv = "../dataset/polar/polar_annotation.csv"
df = pd.read_csv(metadata_path_csv)
```

```python
# build path column
df["path"] = (
    "polar/train_set/" +
    df["patient_id"].astype(str) + "/" +
    df["polyp_id"].astype(str) + "/" +
    df["img_file_prefix"].astype(str) + ".full.png"
)

# rename column
df = df.rename(columns={"polyp_position_(x,y,w,h)": "polyp_pos_xywh"})

# keep only the needed columns from old CSV
df_new = df[["path", "polyp_id", "patient_id", "img_file_prefix", "polyp_pos_xywh", "size", "pathology_diagnosis"]].copy()

# add default annotation fields
df_new["localization_label"]   = -1
df_new["size_label"]           = -1
df_new["cleanliness_label"]    = -1
df_new["overall"]              = -1
df_new["colonexpansion_label"] = -1
df_new["img_cleanliness"]      = -1
df_new["img_overall"]          = -1
df_new["annotator_mail"]       = "none"
```

```python
# save new CSV
output_path = Path("../dataset/db_annotations.csv")
df_new.to_csv(output_path, index=False)

print("✅ db_annotations.csv saved with shape:", df_new.shape)
df_new.head()
```
